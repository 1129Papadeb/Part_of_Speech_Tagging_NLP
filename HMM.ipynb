## Implement a Hidden Markov model in Python

```python
# Hidden Markov model using the training dataset. 

from collections import defaultdict

class HMM:
    def __init__(self):
        self.transition_probs = {} 
        self.emission_probs = {}
        self.tag_counts = defaultdict(int)
        self.tags = set()

    def train(self, tagged_sentences):
        # Use defaultdicts for counting
        transition_counts = defaultdict(lambda: defaultdict(int))
        emission_counts = defaultdict(lambda: defaultdict(int))
        tag_counts = defaultdict(int)

        for sentence in tagged_sentences:
            prev_tag = '<s>'
            tag_counts[prev_tag] += 1

            for word, tag in sentence:
                transition_counts[prev_tag][tag] += 1
                emission_counts[tag][word] += 1
                tag_counts[tag] += 1
                prev_tag = tag

            transition_counts[prev_tag]['</s>'] += 1
            tag_counts['</s>'] += 1

        # Convert counts to probabilities and save as normal dicts.
        self.transition_probs = {
            prev_tag: {
                tag: count / sum(transition_counts[prev_tag].values())
                for tag, count in transition_counts[prev_tag].items()
            }
            for prev_tag in transition_counts
        }

        self.emission_probs = {
            tag: {
                word: count / sum(emission_counts[tag].values())
                for word, count in emission_counts[tag].items()
            }
            for tag in emission_counts
        }

        self.tag_counts = tag_counts
        self.tags = set(tag_counts.keys()) - {'<s>', '</s>'}

    def print_model(self):
        print("Transition Probabilities:")
        for prev_tag in self.transition_probs:
            for tag in self.transition_probs[prev_tag]:
                print(f"P({tag} | {prev_tag}) = {self.transition_probs[prev_tag][tag]:.4f}")

        print("\nEmission Probabilities:")
        for tag in self.emission_probs:
            for word in self.emission_probs[tag]:
                print(f"P({word} | {tag}) = {self.emission_probs[tag][word]:.4f}")
```

```python
dataset = [
    [('The', 'DET'), ('cat', 'NOUN'), ('sleeps', 'VERB')],
    [('A', 'DET'), ('dog', 'NOUN'), ('barks', 'VERB')],
    [('The', 'DET'), ('dog', 'NOUN'), ('sleeps', 'VERB')],
    [('My', 'DET'), ('dog', 'NOUN'), ('runs', 'VERB'), ('fast', 'ADV')],
    [('A', 'DET'), ('cat', 'NOUN'), ('meows', 'VERB'), ('loudly', 'ADV')],
    [('Your', 'DET'), ('cat', 'NOUN'), ('runs', 'VERB')],
    [('The', 'DET'), ('bird', 'NOUN'), ('sings', 'VERB'), ('sweetly', 'ADV')],
    [('A', 'DET'), ('bird', 'NOUN'), ('chirps', 'VERB')],
]
model = HMM()
model.train(dataset)
model.print_model()
```
